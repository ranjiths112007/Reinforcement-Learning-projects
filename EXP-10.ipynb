```python
# EXP-10: Simple cooperative multi-agent experiment
import numpy as np

np.random.seed(31)
positions = np.array([-3.0, 2.0, -2.5])
target = np.zeros(3)

def shared_reward(values):
    return -float(np.sum(np.abs(values - target)))

print(f'Agents: {len(positions)}')
print(f'Initial total distance: {np.sum(np.abs(positions)):.2f}')
for _ in range(3):
    positions += np.clip(-positions, -0.5, 0.5)
print(f'After coordination: {np.sum(np.abs(positions)) + 3.0:.2f}')
print(f'Final total distance: {np.sum(np.abs(positions)):.2f}')
if shared_reward(positions) > -3.0:
    print('Cooperation improved the shared reward.')
```
