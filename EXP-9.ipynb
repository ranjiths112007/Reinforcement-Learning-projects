```python
# EXP-9: Simple model-based control experiment
import numpy as np

np.random.seed(21)

def environment_step(state, action):
    next_state = state + action
    reward = -(next_state ** 2)
    return next_state, reward

state = 2.5
actions = [-0.85, -0.85, -0.85, 0.05]
print(f'Starting state: {state:.2f}')
for step, action in enumerate(actions):
    state, reward = environment_step(state, action)
    print(f'Step {step}: state={state:.2f}, reward={reward:.4f}')
print(f'Final state: {state:.2f}')
```
